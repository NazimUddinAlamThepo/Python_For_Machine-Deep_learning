### Iterating over Dictionary 

```python
dic ={ 
    "name" : "Phitron" , 
    "age": 20 , 
    "address" : "Dhaka" , 
    "numbers":[10,20,30],
    "age":30
    }

keys = dic.keys() # real time update hoy 

print(keys)

dic["math_numebers"] = 30 

print(keys) 

values = dic.values() 

print(values) 

dic["math_numebers"] = 40 
print(values) 

items = dic.items()  # as tuple in a list

print(items)
```

```python
# iteration 
# dictionary iteration er jonno sorboda age .items() e jauya lagbe 
for key,value in dic.items():
    print(key,value)
```

### Dictionary Comprehension

```python
# List comprehension

l = [i**2 for i in range(1,11)  if i%2 == 0]
print(l)
```

```python
square = {x:x**2  for x in range(1,11) if x%2==0}

print(square) 
```

```python
Co_ordinates = [(10,10.5),(20.5,192) , ( 101,102) ] # key

Locations =["dhaka" , "chattogram" , "sylhet"] # value



exact_location = {co_or : loc for co_or, loc in zip(Co_ordinates,Locations)}

print(exact_location) 
```

```python
Co_ordinates = [(10,10.5),(20.5,192) , ( 101,102) ] 

Locations =["dhaka" , "chattogram" , "sylhet"] 



exact_location = {loc : co_or  for co_or, loc in zip(Co_ordinates,Locations)}

print(exact_location) 
```
