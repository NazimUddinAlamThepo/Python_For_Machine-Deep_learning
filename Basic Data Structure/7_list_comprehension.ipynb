```python
even = [] 
# naive
for i in range(101):
    if i%2==0:
        even.append(i)

print(even)
```

```python
# [expression for-loop  if-condition]

l = [i for i in range(10)]
print(l)
```

```python
l = [i for i in range(10)  if i%2 == 0]
print(l)
```
