```python
numbers = [1,2,3,4,5,6]

print(numbers[0:3])
```

## modification 

```python
## adding a item (at first position)
numbers.append(80)
print(numbers)

## insertion at any position 
numbers.insert(0,45)
print(numbers)

## deleting the last value 
numbers.pop()
print(numbers)

## deleting a specific values 
numbers.remove(6)
print(numbers)
```

```python
numbers = [1,2,3,4,5,6,6,6,6,6,6,6,6]

# Delete only one occurance
numbers.remove(6)
print(numbers)
```

### List sorting

```python
numbers.sort()
print(numbers)

```

```python
numbers.reverse()
print(numbers)
```

```python
numbers.sort(reverse = True)
print(numbers)
```
