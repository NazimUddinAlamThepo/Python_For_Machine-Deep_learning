# Module 8 — Object-Oriented Programming (OOP)

Object-Oriented Programming (OOP) is a programming paradigm that organizes code into **objects** — 
bundles of data (attributes) and behavior (methods). Instead of writing a list of instructions, 
you model real-world entities as objects that interact with each other.

> **Why OOP?**  
> Easier to manage large codebases · Promotes code reuse · Mirrors real-world thinking · Easier to debug and extend

## 📦 Section 1: Classes & Objects

### What is a Class?

A **class** is a **blueprint** or **template** for creating objects.  
Think of a class like a cookie cutter — the cutter is the class, each cookie is an object (instance).

```
CLASS (Blueprint)         OBJECT (Instance)
─────────────────         ─────────────────
Phone                →    your_iphone
Car                  →    my_honda
Student              →    alice
```

### What is an Object?

An **object** is a specific instance created from a class.  
Each object has its own data but shares the structure defined by the class.

### Two Core Ingredients of a Class

| Ingredient | Description | Real-World Example |
|---|---|---|
| **Attributes** | Variables that store data | `battery`, `camera`, `color` |
| **Methods** | Functions that define behavior | `charge()`, `take_photo()` |


### 1.1 Creating Your First Class

Syntax:
```python
class ClassName:
    # attributes and methods go here
    pass
```

> **Convention:** Class names use **PascalCase** (e.g., `MyClass`, `BankAccount`, `PhoneBook`)


````python
# Your very first class
class Dog:
    pass  # empty class — no attributes or methods yet

# Creating objects (instances) from the class
dog1 = Dog()
dog2 = Dog()
dog3 = Dog()

print(type(dog1))   # <class '__main__.Dog'>
print(dog1)         # <__main__.Dog object at 0x...>

# dog1, dog2, dog3 are all different objects from the same class
print(dog1 is dog2)  # False — they are different objects
````

### 1.2 Class Attributes vs Instance Attributes

| Type | Where Defined | Shared? | Example |
|---|---|---|---|
| **Class attribute** | Outside `__init__`, inside class | ✅ Yes — all objects share it | `species = "Canis"` |
| **Instance attribute** | Inside `__init__` | ❌ No — each object has its own | `self.name = name` |


````python
# Class attribute example
class Dog:
    species = "Canis lupus familiaris"  # shared by ALL dogs

    def __init__(self, name, breed, age):
        # Instance attributes — unique to each dog
        self.name = name
        self.breed = breed
        self.age = age


# Create objects
dog1 = Dog("Buddy", "Labrador", 3)
dog2 = Dog("Max", "Poodle", 5)
dog3 = Dog("Luna", "Husky", 2)

# Instance attributes — each dog has its own
print(dog1.name, dog1.breed)   # Buddy Labrador
print(dog2.name, dog2.breed)   # Max Poodle

# Class attribute — same for all
print(dog1.species)   # Canis lupus familiaris
print(dog2.species)   # Canis lupus familiaris

# You can override the class attribute on a specific instance
dog1.species = "Mutant Dog"
print(dog1.species)   # Mutant Dog (only dog1 changed)
print(dog2.species)   # Canis lupus familiaris (dog2 unchanged)
````

### 1.3 The Constructor: `__init__`

`__init__` is a **special method** (called a *dunder* — double underscore method) that runs 
**automatically** when you create a new object. It sets up the initial state of the object.

```
dog1 = Dog("Buddy", "Labrador", 3)
         ↑
         Python calls Dog.__init__(dog1, "Buddy", "Labrador", 3) automatically
```

`self` refers to the object being created — it's always the first parameter.


````python
class Student:
    school = "Python Academy"  # class attribute

    def __init__(self, name, age, grade):
        # self = the new student object being created
        self.name = name    # instance attribute
        self.age = age
        self.grade = grade

    def introduce(self):
        print(f"Hi! I'm {self.name}, age {self.age}, in grade {self.grade}.")

    def is_passing(self):
        return self.grade >= 50


# Create students
alice = Student("Alice", 20, 85)
bob   = Student("Bob",   19, 42)
carol = Student("Carol", 21, 91)

# Access attributes
print(alice.name, alice.grade)   # Alice 85

# Call methods
alice.introduce()   # Hi! I'm Alice, age 20, in grade 85.
bob.introduce()     # Hi! I'm Bob, age 19, in grade 42.

# Use method return value
print(f"Alice passing? {alice.is_passing()}")  # True
print(f"Bob passing?   {bob.is_passing()}")    # False
print(f"School: {alice.school}")               # Python Academy
````

### 1.4 Methods — Giving Objects Behavior

Methods are functions inside a class. They always receive `self` as the first argument 
(Python passes it automatically — you don't need to pass it when calling).

**Types of methods you'll see:**

| Method Type | Description | Example |
|---|---|---|
| **Regular method** | Works with instance data via `self` | `def bark(self):` |
| **Method with parameters** | Takes extra arguments | `def eat(self, food):` |
| **Method that returns values** | Uses `return` | `def get_info(self):` |


````python
class Phone:
    manufactured = "China"

    def __init__(self, model, battery_mah, camera_mp, battery_pct=100):
        self.model = model
        self.battery_mah = battery_mah
        self.camera_mp = camera_mp
        self.battery_pct = battery_pct   # default 100%

    # Method: takes extra argument
    def charge(self, minutes):
        """Each minute of charging adds 1% battery (max 100%)."""
        self.battery_pct = min(100, self.battery_pct + minutes)
        print(f"Charged for {minutes} min → {self.battery_pct}%")

    # Method: takes extra argument
    def take_photos(self, count):
        """Each photo costs 1% battery."""
        if self.battery_pct == 0:
            print("❌ Dead battery! Please charge.")
        else:
            drain = min(count, self.battery_pct)   # can't go below 0
            self.battery_pct -= drain
            print(f"Took {drain} photos → {self.battery_pct}% battery left")

    # Method: returns a value
    def get_status(self):
        return f"[{self.model}] 🔋 {self.battery_pct}% | 📷 {self.camera_mp}MP"


# Demo
iphone = Phone("iPhone 17", battery_mah=3500, camera_mp=48)

print(iphone.get_status())   # 100%

iphone.take_photos(60)       # 60 photos → 40%
iphone.take_photos(50)       # 40 more (runs out) → 0%
iphone.take_photos(5)        # dead battery message

iphone.charge(30)            # charge 30 min → 30%
print(iphone.get_status())
````

### 1.5 Real-World Example: Bank Account

Let's build a more complete example that shows how classes model real-world entities.


````python
class BankAccount:
    bank_name = "Python Bank"   # class attribute

    def __init__(self, owner, initial_balance=0):
        self.owner = owner
        self.balance = initial_balance
        self.transactions = []   # list of transaction records

    def deposit(self, amount):
        if amount <= 0:
            print("❌ Deposit amount must be positive.")
            return
        self.balance += amount
        self.transactions.append(f"+{amount}")
        print(f"✅ Deposited {amount}. Balance: {self.balance}")

    def withdraw(self, amount):
        if amount <= 0:
            print("❌ Withdrawal amount must be positive.")
        elif amount > self.balance:
            print(f"❌ Insufficient funds. Balance: {self.balance}")
        else:
            self.balance -= amount
            self.transactions.append(f"-{amount}")
            print(f"✅ Withdrew {amount}. Balance: {self.balance}")

    def get_statement(self):
        print(f"\n{'='*35}")
        print(f" {self.bank_name} | {self.owner}'s Account")
        print(f"{'='*35}")
        for t in self.transactions:
            print(f"  {t}")
        print(f"  Current Balance: {self.balance}")
        print(f"{'='*35}\n")


# Create accounts
alice_acc = BankAccount("Alice", initial_balance=1000)
bob_acc   = BankAccount("Bob")

alice_acc.deposit(500)
alice_acc.withdraw(200)
alice_acc.withdraw(2000)   # will fail — insufficient
alice_acc.get_statement()

bob_acc.deposit(300)
bob_acc.withdraw(50)
bob_acc.get_statement()
````

## 🧬 Section 2: Inheritance

### What is Inheritance?

**Inheritance** allows a class (child) to **reuse** code from another class (parent).  
The child gets all attributes and methods of the parent **for free**, and can add its own.

```
Animal (Parent / Base class)
  ├── Dog   (Child / Derived class)
  ├── Cat   (Child)
  └── Bird  (Child)
```

**Why inheritance?**
- Avoid repeating the same code in multiple classes (DRY principle — Don't Repeat Yourself)
- Create a natural hierarchy that mirrors the real world
- Easy to extend — add new child classes without modifying the parent

**Syntax:**
```python
class Child(Parent):
    ...
```


````python
# Parent class
class Animal:
    alive = True   # class attribute

    def __init__(self, name, sound):
        self.name = name
        self.sound = sound

    def eat(self):
        print(f"{self.name} is eating.")

    def sleep(self):
        print(f"{self.name} is sleeping.")

    def make_sound(self):
        print(f"{self.name} says: {self.sound}")


# Child classes — inherit everything from Animal
class Dog(Animal):
    pass   # no extra code needed — Dog gets all Animal methods


class Cat(Animal):
    pass


class Bird(Animal):
    pass


# Each child has Animal's methods
dog  = Dog("Buddy", "Woof!")
cat  = Cat("Whiskers", "Meow!")
bird = Bird("Tweety", "Tweet!")

dog.eat()         # Buddy is eating.
cat.sleep()       # Whiskers is sleeping.
bird.make_sound() # Tweety says: Tweet!

print(dog.alive)  # True — inherited class attribute
````

### 2.1 Adding Child-Specific Attributes and Methods

Children can have **extra** attributes and methods beyond what they inherit.


````python
class Animal:
    def __init__(self, name):
        self.name = name

    def eat(self):
        print(f"{self.name} is eating.")

    def info(self):
        print(f"Animal name: {self.name}")


class Dog(Animal):
    def __init__(self, name, breed):
        self.name = name     # setting parent's attribute manually
        self.breed = breed   # Dog-specific attribute

    def fetch(self):
        print(f"{self.name} fetches the ball! 🎾")

    def info(self):          # overriding parent's info method
        print(f"Dog name: {self.name} | Breed: {self.breed}")


class Bird(Animal):
    def __init__(self, name, can_fly):
        self.name = name
        self.can_fly = can_fly   # Bird-specific attribute

    def fly(self):
        if self.can_fly:
            print(f"{self.name} is soaring through the sky! 🦅")
        else:
            print(f"{self.name} cannot fly. 🐧")

    def info(self):
        print(f"Bird name: {self.name} | Can fly: {self.can_fly}")


# Demo
rex   = Dog("Rex", "German Shepherd")
parrot = Bird("Polly", can_fly=True)
penguin = Bird("Pingu", can_fly=False)

rex.eat()         # inherited from Animal
rex.fetch()       # Dog-specific
rex.info()        # overridden method

parrot.eat()      # inherited
parrot.fly()      # Bird-specific
parrot.info()

penguin.fly()     # "cannot fly"
````

### 2.2 `super()` — Reusing the Parent Constructor

Instead of rewriting `self.name = name` in every child class, use `super().__init__()` 
to call the parent constructor automatically.

```
super().__init__(args)
       ↑
       Calls the parent class's __init__ method
```


````python
# ❌ Without super() — repetitive code
class Shape:
    def __init__(self, color, filled):
        self.color = color
        self.filled = filled

class Circle_Bad:
    def __init__(self, color, filled, radius):
        self.color = color    # repeated!
        self.filled = filled  # repeated!
        self.radius = radius

class Square_Bad:
    def __init__(self, color, filled, side):
        self.color = color    # repeated!
        self.filled = filled  # repeated!
        self.side = side


# ✅ With super() — clean and DRY
class Shape:
    def __init__(self, color, filled):
        self.color = color
        self.filled = filled

    def describe(self):
        state = "filled" if self.filled else "not filled"
        print(f"A {self.color}, {state} shape.")


class Circle(Shape):
    def __init__(self, color, filled, radius):
        super().__init__(color, filled)   # ← hands off to Shape.__init__
        self.radius = radius

    def area(self):
        return round(3.14159 * self.radius ** 2, 2)


class Square(Shape):
    def __init__(self, color, filled, side):
        super().__init__(color, filled)
        self.side = side

    def area(self):
        return self.side ** 2


class Triangle(Shape):
    def __init__(self, color, filled, base, height):
        super().__init__(color, filled)
        self.base = base
        self.height = height

    def area(self):
        return 0.5 * self.base * self.height


# Demo
c = Circle("red", True, radius=5)
s = Square("blue", False, side=4)
t = Triangle("green", True, base=6, height=3)

c.describe()
print(f"Circle area: {c.area()}")

s.describe()
print(f"Square area: {s.area()}")

t.describe()
print(f"Triangle area: {t.area()}")
````

### 2.3 Method Overriding

A child class can **redefine** a method from the parent. When you call that method on 
the child object, Python uses the child's version.

```
child_object.method()
       ↑
       Python looks for method in:  Child → Parent → Grandparent → ...
       (This is called Method Resolution Order, or MRO)
```


````python
class Employee:
    def __init__(self, name, base_salary):
        self.name = name
        self.base_salary = base_salary

    def get_pay(self):
        return self.base_salary

    def __str__(self):
        return f"{self.name} — Pay: ${self.get_pay():,}"


class Manager(Employee):
    def __init__(self, name, base_salary, bonus):
        super().__init__(name, base_salary)
        self.bonus = bonus

    def get_pay(self):             # overrides Employee.get_pay()
        return self.base_salary + self.bonus


class Intern(Employee):
    def __init__(self, name, base_salary):
        super().__init__(name, base_salary)

    def get_pay(self):             # overrides with 80% of salary
        return self.base_salary * 0.8


# Demo
emp     = Employee("Alice", 5000)
manager = Manager("Bob", 5000, bonus=2000)
intern  = Intern("Carol", 2000)

staff = [emp, manager, intern]

for person in staff:
    print(person)
````

### 2.4 Extending a Parent Method with `super()`

Instead of completely replacing the parent method, you can **call it first** and then add more behavior.


````python
class Shape:
    def __init__(self, color, filled):
        self.color = color
        self.filled = filled

    def describe(self):
        state = "filled" if self.filled else "not filled"
        print(f"Shape → Color: {self.color} | {state}")


class Circle(Shape):
    PI = 3.14159

    def __init__(self, color, filled, radius):
        super().__init__(color, filled)
        self.radius = radius

    def describe(self):
        super().describe()                          # first call parent
        area = self.PI * self.radius ** 2
        print(f"Circle → Radius: {self.radius} cm | Area: {area:.2f} cm²")


class Square(Shape):
    def __init__(self, color, filled, side):
        super().__init__(color, filled)
        self.side = side

    def describe(self):
        super().describe()
        print(f"Square → Side: {self.side} cm | Area: {self.side**2} cm²")


c = Circle("red", True, 7)
c.describe()

print()

s = Square("blue", False, 4)
s.describe()
````

### 2.5 Multi-Level Inheritance

Inheritance can chain across multiple levels: Grandparent → Parent → Child.


````python
# Multi-level inheritance
class LivingThing:
    def breathe(self):
        print("Breathing...")


class Animal(LivingThing):
    def __init__(self, name):
        self.name = name

    def eat(self):
        print(f"{self.name} is eating.")


class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)
        self.breed = breed

    def bark(self):
        print(f"{self.name} says: Woof! 🐶")


# Dog inherits from Animal, which inherits from LivingThing
rex = Dog("Rex", "Labrador")

rex.breathe()   # from LivingThing (grandparent)
rex.eat()       # from Animal (parent)
rex.bark()      # from Dog (own)

# Check inheritance chain
print(Dog.__mro__)   # Method Resolution Order
````

### 2.6 Multiple Inheritance

A class can inherit from **more than one parent**.

> ⚠️ When parents have **different constructor signatures**, call them explicitly instead of using `super()`.


````python
# Multiple inheritance with different constructor signatures
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        print(f"Hi, I'm {self.name}, age {self.age}.")


class Worker:
    def __init__(self, job_title, salary):
        self.job_title = job_title
        self.salary = salary

    def work(self):
        print(f"Working as a {self.job_title}.")


class Employee(Person, Worker):
    def __init__(self, name, age, job_title, salary, employee_id):
        Person.__init__(self, name, age)          # call explicitly
        Worker.__init__(self, job_title, salary)  # call explicitly
        self.employee_id = employee_id

    def profile(self):
        print(f"ID: {self.employee_id} | {self.name} | {self.job_title} | Salary: ${self.salary:,}")


emp = Employee("Alice", 30, "Software Engineer", 85000, "E-1042")

emp.introduce()   # from Person
emp.work()        # from Worker
emp.profile()     # own method
````

## 🎭 Section 3: Polymorphism

### What is Polymorphism?

**Polymorphism** means *"many forms"* — the same method name behaves differently 
depending on which object calls it.

Think of a universal remote. The **Power** button does the same *concept* (turn on/off), 
but the actual behavior differs for TV, AC, fan, etc.

```
animal.speak()
  Dog  →  "Woof!"
  Cat  →  "Meow!"
  Cow  →  "Moo!"
```

All objects respond to the same method name, but each does it its own way.

**Two types of polymorphism in Python:**
1. **Method Overriding** — Child redefines a parent method  
2. **Duck Typing** — "If it walks like a duck and quacks like a duck, it's a duck" — Python doesn't care about the class, just whether the method exists


````python
# Method overriding polymorphism
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return "..."   # default — subclasses should override


class Dog(Animal):
    def speak(self):
        return "Woof! 🐶"


class Cat(Animal):
        def speak(self):
            return "Meow! 🐱"


class Cow(Animal):
    def speak(self):
        return "Moo! 🐄"


class Snake(Animal):
    def speak(self):
        return "Hiss! 🐍"


# Polymorphism in action — same method, different behavior
animals = [
    Dog("Buddy"),
    Cat("Whiskers"),
    Cow("Bessie"),
    Snake("Sly")
]

for animal in animals:
    print(f"{animal.name:10} → {animal.speak()}")
````

````python
# Duck typing — works on any object that has the right method
class Robot:
    def speak(self):
        return "Beep boop! 🤖"


class Parrot:
    def speak(self):
        return "Polly wants a cracker! 🦜"


class Baby:
    def speak(self):
        return "Goo goo ga ga! 👶"


# Robot, Parrot, Baby are NOT related — no common parent
# But they all have speak() → polymorphism via duck typing
speakers = [Robot(), Parrot(), Baby()]

for speaker in speakers:
    # Python doesn't check the class — just calls .speak()
    print(f"{type(speaker).__name__:8}: {speaker.speak()}")
````

### 3.1 Polymorphism with Geometry — Real Example

A common pattern: a parent class defines the *interface* (what methods exist), 
and each child implements the *behavior* differently.


````python
import math

class Shape:
    def __init__(self, color="black"):
        self.color = color

    def area(self):
        raise NotImplementedError("Subclass must implement area()")

    def perimeter(self):
        raise NotImplementedError("Subclass must implement perimeter()")

    def describe(self):
        print(f"{type(self).__name__:12} | Color: {self.color:8} | "
              f"Area: {self.area():8.2f} | Perimeter: {self.perimeter():.2f}")


class Circle(Shape):
    def __init__(self, radius, color="black"):
        super().__init__(color)
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

    def perimeter(self):
        return 2 * math.pi * self.radius


class Rectangle(Shape):
    def __init__(self, width, height, color="black"):
        super().__init__(color)
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)


class Triangle(Shape):
    def __init__(self, a, b, c, color="black"):
        super().__init__(color)
        self.a, self.b, self.c = a, b, c

    def area(self):
        s = (self.a + self.b + self.c) / 2     # semi-perimeter
        return math.sqrt(s * (s-self.a) * (s-self.b) * (s-self.c))

    def perimeter(self):
        return self.a + self.b + self.c


# All shapes, one loop
shapes = [
    Circle(5, "red"),
    Rectangle(4, 6, "blue"),
    Triangle(3, 4, 5, "green"),
    Circle(10, "yellow"),
    Rectangle(2, 8, "purple"),
]

print(f"{'Shape':12} | {'Color':8} | {'Area':>8} | {'Perimeter'}")
print("-" * 55)
for shape in shapes:
    shape.describe()

# Find total area
total = sum(s.area() for s in shapes)
print(f"\nTotal area of all shapes: {total:.2f}")
````

### 3.2 Abstract Base Classes — Enforcing the Contract

`ABC` (Abstract Base Class) + `@abstractmethod` forces child classes to implement required methods.  
If a child forgets, Python raises an error **at object creation time** (not later when the method is called).


````python
from abc import ABC, abstractmethod


class Vehicle(ABC):
    """
    Abstract base class — defines the INTERFACE every vehicle must have.
    You cannot create a Vehicle() object directly.
    """

    def __init__(self, brand, speed_kmh):
        self.brand = brand
        self.speed_kmh = speed_kmh

    @abstractmethod
    def start(self):
        """Must be implemented by every child class."""
        pass

    @abstractmethod
    def stop(self):
        """Must be implemented by every child class."""
        pass

    # Concrete method (not abstract) — shared by all
    def info(self):
        return f"{self.brand} | Max speed: {self.speed_kmh} km/h"


class Car(Vehicle):
    def start(self): return f"🚗 {self.brand}: Engine started. Vroom!"
    def stop(self):  return f"🚗 {self.brand}: Brakes applied. Stopped."


class ElectricCar(Vehicle):
    def start(self): return f"⚡ {self.brand}: Silently powered on."
    def stop(self):  return f"⚡ {self.brand}: Regenerative braking. Stopped."


class Bicycle(Vehicle):
    def start(self): return f"🚲 {self.brand}: Pedaling started."
    def stop(self):  return f"🚲 {self.brand}: Hand brakes applied."


class Boat(Vehicle):
    def start(self): return f"⛵ {self.brand}: Engine ignited. Sailing!"
    def stop(self):  return f"⛵ {self.brand}: Anchor dropped."


# Try creating abstract class directly → error
try:
    v = Vehicle("Generic", 100)
except TypeError as e:
    print(f"Error: {e}\n")


# Polymorphism with all vehicles
fleet = [
    Car("Toyota", 180),
    ElectricCar("Tesla", 250),
    Bicycle("Trek", 30),
    Boat("Yamaha", 60),
]

for vehicle in fleet:
    print(vehicle.info())
    print(" ", vehicle.start())
    print(" ", vehicle.stop())
    print()
````

````python
# Practical example: Notification system
from abc import ABC, abstractmethod


class Notifier(ABC):
    """Abstract notifier — any new channel just implements send()."""

    def __init__(self, recipient):
        self.recipient = recipient

    @abstractmethod
    def send(self, message):
        pass

    def notify(self, message):
        print(f"Sending to {self.recipient}...")
        self.send(message)
        print("Done!\n")


class EmailNotifier(Notifier):
    def send(self, message):
        print(f"📧 Email → {self.recipient}: {message}")


class SMSNotifier(Notifier):
    def send(self, message):
        print(f"📱 SMS → {self.recipient}: {message}")


class PushNotifier(Notifier):
    def send(self, message):
        print(f"🔔 Push → {self.recipient}: {message}")


class SlackNotifier(Notifier):
    def send(self, message):
        print(f"💬 Slack → {self.recipient}: {message}")


# Use same interface for all channels
notifiers = [
    EmailNotifier("alice@email.com"),
    SMSNotifier("+8801700000000"),
    PushNotifier("device_token_abc123"),
    SlackNotifier("#general"),
]

for n in notifiers:
    n.notify("Your order has been shipped! 📦")
````

## 🔒 Section 4: Encapsulation

### What is Encapsulation?

**Encapsulation** means **bundling data (attributes) and the methods that operate on that data 
inside one class**, AND **restricting direct access** to some of the object's components.

**Why?**  
- Protect internal data from accidental corruption  
- Control *how* data is accessed and modified  
- Hide implementation details — users only see what they need

**Real-world analogy:**  
A capsule (medicine) — you take the pill, but you don't touch the chemical inside.  
An ATM — you interact through the screen/buttons; you can't access the cash directly.

### Access Modifiers in Python

| Modifier | Syntax | Meaning |
|---|---|---|
| **Public** | `self.name` | Anyone can access |
| **Protected** | `self._name` | Convention: for internal/subclass use |
| **Private** | `self.__name` | Name-mangled — hard to access from outside |

> Python doesn't *enforce* access restrictions like Java/C++.  
> It uses **naming conventions** and **name mangling** to signal intent.


````python
# Demonstrating all three access levels
class Person:
    def __init__(self, name, _id_number, __password):
        self.name = name           # public
        self._id_number = _id_number    # protected
        self.__password = __password   # private (name-mangled)

    def get_info(self):
        # Inside the class, all are accessible
        print(f"Name: {self.name}")
        print(f"ID: {self._id_number}")
        print(f"Password: {self.__password}")


p = Person("Alice", "ID-001", "secret123")

# Public — works anywhere
print(p.name)            # Alice

# Protected — works but convention says "be careful"
print(p._id_number)      # ID-001

# Private — name-mangled to _Person__password
try:
    print(p.__password)  # ❌ AttributeError
except AttributeError as e:
    print(f"Error: {e}")

# Still accessible via mangled name (not recommended)
print(p._Person__password)    # secret123  ← use only for debugging

# Best practice: access private data through methods
p.get_info()
````

### 4.1 Getters and Setters — Controlled Access

Instead of exposing private attributes directly, provide **getter** (read) and **setter** (write) methods.  
This lets you add **validation logic** when reading or writing data.


````python
class Student:
    def __init__(self, name, age, grade):
        self.name = name
        self.__age = age       # private
        self.__grade = grade   # private

    # Getter — read the private value
    def get_age(self):
        return self.__age

    # Setter — validate before setting
    def set_age(self, age):
        if not isinstance(age, int):
            print("❌ Age must be an integer.")
        elif age < 0 or age > 150:
            print("❌ Age must be between 0 and 150.")
        else:
            self.__age = age
            print(f"✅ Age updated to {self.__age}")

    def get_grade(self):
        return self.__grade

    def set_grade(self, grade):
        if 0 <= grade <= 100:
            self.__grade = grade
            print(f"✅ Grade updated to {self.__grade}")
        else:
            print("❌ Grade must be between 0 and 100.")

    def report(self):
        status = "Pass ✅" if self.__grade >= 50 else "Fail ❌"
        print(f"Student: {self.name} | Age: {self.__age} | Grade: {self.__grade} | {status}")


s = Student("Alice", 20, 85)
s.report()

s.set_age(21)       # ✅ valid
s.set_age(-5)       # ❌ invalid
s.set_age("twenty") # ❌ invalid

s.set_grade(95)     # ✅
s.set_grade(150)    # ❌

s.report()
````

### 4.2 Python `@property` — Elegant Getters and Setters

Python has a built-in `@property` decorator that makes getters/setters look like regular attribute access — cleaner and more Pythonic.


````python
class Temperature:
    def __init__(self, celsius):
        self.__celsius = celsius   # private

    # @property makes this a getter — called like temp.celsius (no parentheses)
    @property
    def celsius(self):
        return self.__celsius

    # @celsius.setter — called like temp.celsius = 25
    @celsius.setter
    def celsius(self, value):
        if value < -273.15:
            raise ValueError("Temperature below absolute zero!")
        self.__celsius = value

    # Computed property — converts automatically
    @property
    def fahrenheit(self):
        return self.__celsius * 9/5 + 32

    @property
    def kelvin(self):
        return self.__celsius + 273.15

    def __str__(self):
        return f"{self.__celsius}°C = {self.fahrenheit:.1f}°F = {self.kelvin:.2f}K"


temp = Temperature(25)
print(temp)           # 25°C = 77.0°F = 298.15K

temp.celsius = 100    # uses the setter — validation included
print(temp)           # 100°C = ...

try:
    temp.celsius = -300   # below absolute zero → error
except ValueError as e:
    print(f"Error: {e}")

# Read-only property (no setter)
print(f"Boiling point in Fahrenheit: {temp.fahrenheit}")
````

### 4.3 Full Encapsulation Example: Inventory System


````python
class Product:
    def __init__(self, name, price, quantity):
        self.__name = name
        self.__price = price
        self.__quantity = quantity
        self.__sales = 0

    @property
    def name(self):
        return self.__name

    @property
    def price(self):
        return self.__price

    @price.setter
    def price(self, value):
        if value < 0:
            print("❌ Price cannot be negative.")
        else:
            self.__price = value

    @property
    def quantity(self):
        return self.__quantity

    def restock(self, amount):
        if amount > 0:
            self.__quantity += amount
            print(f"✅ Restocked {amount} units of '{self.__name}'. Total: {self.__quantity}")
        else:
            print("❌ Restock amount must be positive.")

    def sell(self, amount):
        if amount <= 0:
            print("❌ Sale amount must be positive.")
        elif amount > self.__quantity:
            print(f"❌ Only {self.__quantity} units available.")
        else:
            self.__quantity -= amount
            self.__sales += amount
            revenue = amount * self.__price
            print(f"✅ Sold {amount} × '{self.__name}' = ${revenue:.2f}")

    def summary(self):
        print(f"{'─'*40}")
        print(f"  Product  : {self.__name}")
        print(f"  Price    : ${self.__price:.2f}")
        print(f"  In Stock : {self.__quantity}")
        print(f"  Total Sold: {self.__sales}")
        print(f"  Revenue  : ${self.__sales * self.__price:.2f}")
        print(f"{'─'*40}")


laptop = Product("Laptop", 999.99, 50)
laptop.sell(10)
laptop.sell(5)
laptop.restock(20)
laptop.price = 899.99   # price drop via setter
laptop.sell(3)
laptop.summary()
````

## 🎩 Section 5: Abstraction

### What is Abstraction?

**Abstraction** means hiding the **complex internal implementation** and showing only the **essential interface**.

You deal with abstractions every day:
- You press the car's accelerator — you don't need to understand combustion
- You call `print()` — you don't need to know how the terminal renders text
- You use a coffee machine — you press *start*, not manually control the boiler

> **Encapsulation** = hiding *data*  
> **Abstraction** = hiding *complexity*

### How to Achieve Abstraction in Python

Using `ABC` (Abstract Base Class) from the `abc` module:
1. Inherit from `ABC`
2. Mark methods with `@abstractmethod`
3. Child classes **must** implement those methods


````python
from abc import ABC, abstractmethod


# Abstract class — defines WHAT to do, not HOW
class DatabaseConnector(ABC):

    @abstractmethod
    def connect(self):
        pass

    @abstractmethod
    def disconnect(self):
        pass

    @abstractmethod
    def execute_query(self, query):
        pass

    # Concrete method — shared implementation
    def run(self, query):
        self.connect()
        result = self.execute_query(query)
        self.disconnect()
        return result


# Concrete implementations — define HOW
class MySQLConnector(DatabaseConnector):
    def connect(self):
        print("🐬 Connected to MySQL database.")

    def disconnect(self):
        print("🐬 Disconnected from MySQL.")

    def execute_query(self, query):
        print(f"🐬 MySQL executing: {query}")
        return ["row1", "row2"]   # simulated result


class PostgreSQLConnector(DatabaseConnector):
    def connect(self):
        print("🐘 Connected to PostgreSQL database.")

    def disconnect(self):
        print("🐘 Disconnected from PostgreSQL.")

    def execute_query(self, query):
        print(f"🐘 PostgreSQL executing: {query}")
        return ["result_a", "result_b"]


class SQLiteConnector(DatabaseConnector):
    def connect(self):
        print("🗄️  Connected to SQLite file.")

    def disconnect(self):
        print("🗄️  Closed SQLite file.")

    def execute_query(self, query):
        print(f"🗄️  SQLite executing: {query}")
        return ["local_data"]


# User just calls .run() — doesn't care about the internals
databases = [MySQLConnector(), PostgreSQLConnector(), SQLiteConnector()]

for db in databases:
    result = db.run("SELECT * FROM users")
    print(f"Result: {result}\n")
````

````python
# Abstraction example: Payment processing
from abc import ABC, abstractmethod


class PaymentProcessor(ABC):
    """
    Abstract class — defines the payment interface.
    New payment methods are added WITHOUT changing existing code.
    """

    def __init__(self, amount):
        self.amount = amount
        self._transaction_id = None

    @abstractmethod
    def validate(self):
        """Check if payment details are valid."""
        pass

    @abstractmethod
    def process(self):
        """Execute the payment."""
        pass

    @abstractmethod
    def get_receipt(self):
        """Return a receipt string."""
        pass

    # Template method — concrete steps using abstract methods
    def pay(self):
        print(f"{'─'*40}")
        if self.validate():
            self.process()
            print(self.get_receipt())
        else:
            print("❌ Payment validation failed.")
        print(f"{'─'*40}\n")


class CreditCardPayment(PaymentProcessor):
    def __init__(self, amount, card_number):
        super().__init__(amount)
        self.card_number = card_number

    def validate(self):
        # simplified: card must be 16 digits
        return len(str(self.card_number)) == 16

    def process(self):
        self._transaction_id = f"CC-{hash(self.card_number) % 100000:05d}"
        print(f"💳 Processing credit card payment of ${self.amount}...")

    def get_receipt(self):
        return f"✅ Credit card payment of ${self.amount} approved. Txn: {self._transaction_id}"


class BKashPayment(PaymentProcessor):
    def __init__(self, amount, phone):
        super().__init__(amount)
        self.phone = phone

    def validate(self):
        return str(self.phone).startswith("01") and len(str(self.phone)) == 11

    def process(self):
        self._transaction_id = f"BK-{hash(self.phone) % 100000:05d}"
        print(f"📱 Processing bKash payment of ৳{self.amount}...")

    def get_receipt(self):
        return f"✅ bKash payment of ৳{self.amount} successful. Txn: {self._transaction_id}"


class CryptoPayment(PaymentProcessor):
    def __init__(self, amount, wallet_address):
        super().__init__(amount)
        self.wallet = wallet_address

    def validate(self):
        return len(self.wallet) >= 26   # simplified wallet check

    def process(self):
        self._transaction_id = f"BTC-{hash(self.wallet) % 100000:05d}"
        print(f"₿  Broadcasting crypto transaction of ${self.amount}...")

    def get_receipt(self):
        return f"✅ Crypto payment of ${self.amount} confirmed. Txn: {self._transaction_id}"


# All handled through the same interface
payments = [
    CreditCardPayment(150, 4111111111111111),
    BKashPayment(500, "01712345678"),
    CryptoPayment(75, "1A1zP1eP5QGefi2DMPTfTL5SLmv7Divf"),
    BKashPayment(200, "123"),  # invalid — too short
]

for payment in payments:
    payment.pay()
````

````python
# Simple abstraction example — shapes with abstract area()
from abc import ABC, abstractmethod
import math


class Shape(ABC):
    """Every shape MUST know how to compute its area and perimeter."""

    @abstractmethod
    def area(self) -> float:
        pass

    @abstractmethod
    def perimeter(self) -> float:
        pass

    # Concrete — available to all shapes
    def print_info(self):
        print(f"{type(self).__name__:12} → Area: {self.area():8.2f} | Perimeter: {self.perimeter():.2f}")


class Circle(Shape):
    def __init__(self, r): self.r = r
    def area(self): return math.pi * self.r ** 2
    def perimeter(self): return 2 * math.pi * self.r


class Rectangle(Shape):
    def __init__(self, w, h): self.w, self.h = w, h
    def area(self): return self.w * self.h
    def perimeter(self): return 2 * (self.w + self.h)


class Square(Shape):
    def __init__(self, s): self.s = s
    def area(self): return self.s ** 2
    def perimeter(self): return 4 * self.s


class Ellipse(Shape):
    def __init__(self, a, b): self.a, self.b = a, b
    def area(self): return math.pi * self.a * self.b
    def perimeter(self): return 2 * math.pi * math.sqrt((self.a**2 + self.b**2) / 2)


shapes = [Circle(5), Rectangle(4, 6), Square(3), Ellipse(5, 3)]

print(f"{'Shape':12} → {'Area':>8} | {'Perimeter'}")
print("─" * 45)
for s in shapes:
    s.print_info()

largest = max(shapes, key=lambda s: s.area())
print(f"\nLargest shape: {type(largest).__name__} with area {largest.area():.2f}")
````

## 🏗️ Section 6: OOP in Practice — All Pillars Together

The real power of OOP comes when all pillars work together.  
Let's build a mini **Library Management System** that demonstrates every concept.


````python
from abc import ABC, abstractmethod


# ── Abstraction: define what every library item must do ──────────
class LibraryItem(ABC):
    total_items = 0   # Class attribute (shared counter)

    def __init__(self, title, item_id):
        self.__title = title       # Encapsulation: private
        self.__item_id = item_id
        self._is_borrowed = False  # protected
        LibraryItem.total_items += 1

    # Abstraction: child must implement these
    @abstractmethod
    def get_type(self):
        pass

    @abstractmethod
    def get_details(self):
        pass

    # Encapsulation: controlled access via properties
    @property
    def title(self):
        return self.__title

    @property
    def item_id(self):
        return self.__item_id

    @property
    def is_borrowed(self):
        return self._is_borrowed

    def borrow(self):
        if self._is_borrowed:
            print(f"❌ '{self.title}' is already borrowed.")
        else:
            self._is_borrowed = True
            print(f"✅ '{self.title}' borrowed successfully.")

    def return_item(self):
        if not self._is_borrowed:
            print(f"⚠️  '{self.title}' was not borrowed.")
        else:
            self._is_borrowed = False
            print(f"✅ '{self.title}' returned. Thank you!")

    def status(self):
        state = "🔴 Borrowed" if self._is_borrowed else "🟢 Available"
        return f"[{self.item_id}] {self.get_type():10} | {state} | {self.title}"


# ── Inheritance: specific item types ────────────────────────────
class Book(LibraryItem):
    def __init__(self, title, item_id, author, pages):
        super().__init__(title, item_id)
        self.__author = author  # private
        self.__pages = pages

    def get_type(self):
        return "Book"

    def get_details(self):
        print(f"📚 Book   : {self.title}")
        print(f"   Author : {self.__author}")
        print(f"   Pages  : {self.__pages}")
        print(f"   Status : {'Borrowed' if self.is_borrowed else 'Available'}")


class DVD(LibraryItem):
    def __init__(self, title, item_id, director, duration_min):
        super().__init__(title, item_id)
        self.__director = director
        self.__duration = duration_min

    def get_type(self):
        return "DVD"

    def get_details(self):
        print(f"🎬 DVD      : {self.title}")
        print(f"   Director : {self.__director}")
        print(f"   Duration : {self.__duration} min")
        print(f"   Status   : {'Borrowed' if self.is_borrowed else 'Available'}")


class Magazine(LibraryItem):
    def __init__(self, title, item_id, issue_number):
        super().__init__(title, item_id)
        self.__issue = issue_number

    def get_type(self):
        return "Magazine"

    def get_details(self):
        print(f"📰 Magazine : {self.title}")
        print(f"   Issue    : #{self.__issue}")
        print(f"   Status   : {'Borrowed' if self.is_borrowed else 'Available'}")


# ── Library class ────────────────────────────────────────────────
class Library:
    def __init__(self, name):
        self.__name = name
        self.__items = {}   # private — dict of item_id → LibraryItem

    def add_item(self, item):
        self.__items[item.item_id] = item
        print(f"Added: {item.status()}")

    def borrow_item(self, item_id):
        item = self.__items.get(item_id)
        if item:
            item.borrow()
        else:
            print(f"❌ Item {item_id} not found.")

    def return_item(self, item_id):
        item = self.__items.get(item_id)
        if item:
            item.return_item()
        else:
            print(f"❌ Item {item_id} not found.")

    def catalog(self):
        print(f"\n{'═'*55}")
        print(f"  📖 {self.__name} — Catalog ({len(self.__items)} items)")
        print(f"{'═'*55}")
        for item in self.__items.values():
            print(" ", item.status())
        print(f"{'═'*55}\n")


# ── Demo ─────────────────────────────────────────────────────────
lib = Library("City Central Library")
print()

# Add items
lib.add_item(Book("Clean Code", "B001", "Robert C. Martin", 431))
lib.add_item(Book("Python Crash Course", "B002", "Eric Matthes", 544))
lib.add_item(DVD("Inception", "D001", "Christopher Nolan", 148))
lib.add_item(Magazine("National Geographic", "M001", 245))
lib.add_item(Book("The Pragmatic Programmer", "B003", "Hunt & Thomas", 352))

lib.catalog()

# Borrow and return
lib.borrow_item("B001")
lib.borrow_item("D001")
lib.borrow_item("B001")   # already borrowed

lib.catalog()

lib.return_item("B001")
lib.catalog()

print(f"Total library items ever created: {LibraryItem.total_items}")
````

## ⚡ Bonus: Special (Dunder) Methods

Python classes have **special methods** surrounded by double underscores (`__`).  
They let your objects behave like built-in types (support `print()`, `+`, `==`, `len()`, etc.).

| Method | Triggered By | Purpose |
|---|---|---|
| `__init__` | `MyClass()` | Constructor |
| `__str__` | `print(obj)`, `str(obj)` | Human-readable string |
| `__repr__` | `repr(obj)` | Debug representation |
| `__len__` | `len(obj)` | Length |
| `__add__` | `obj1 + obj2` | Addition operator |
| `__eq__` | `obj1 == obj2` | Equality check |
| `__lt__` | `obj1 < obj2` | Less-than comparison |


````python
class Vector:
    """A 2D vector that supports math operations naturally."""

    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        return f"Vector({self.x}, {self.y})"

    def __repr__(self):
        return f"Vector(x={self.x}, y={self.y})"

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __sub__(self, other):
        return Vector(self.x - other.x, self.y - other.y)

    def __mul__(self, scalar):
        return Vector(self.x * scalar, self.y * scalar)

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y

    def __len__(self):
        import math
        return int(math.sqrt(self.x**2 + self.y**2))   # magnitude

    def magnitude(self):
        import math
        return math.sqrt(self.x**2 + self.y**2)


v1 = Vector(3, 4)
v2 = Vector(1, 2)

print(v1)           # uses __str__  → Vector(3, 4)
print(repr(v1))     # uses __repr__ → Vector(x=3, y=4)
print(v1 + v2)      # uses __add__  → Vector(4, 6)
print(v1 - v2)      # uses __sub__  → Vector(2, 2)
print(v1 * 3)       # uses __mul__  → Vector(9, 12)
print(v1 == v2)     # uses __eq__   → False
print(v1 == Vector(3, 4))  # True

print(f"Magnitude of {v1}: {v1.magnitude():.2f}")   # 5.00
````

````python
class Playlist:
    """A music playlist with dunder methods."""

    def __init__(self, name):
        self.name = name
        self.__songs = []

    def add(self, song):
        self.__songs.append(song)

    def __str__(self):
        return f"Playlist: '{self.name}' ({len(self.__songs)} songs)"

    def __len__(self):
        return len(self.__songs)

    def __contains__(self, song):
        return song in self.__songs

    def __getitem__(self, index):
        return self.__songs[index]

    def __iter__(self):
        return iter(self.__songs)


pl = Playlist("Chill Vibes")
pl.add("Blinding Lights")
pl.add("Levitating")
pl.add("Stay")
pl.add("Starboy")

print(pl)                       # uses __str__
print(f"Length: {len(pl)}")     # uses __len__

print("Levitating" in pl)       # uses __contains__  → True
print("Bohemian Rhapsody" in pl)  # False

print(f"First song: {pl[0]}")   # uses __getitem__

print("\nAll songs:")
for song in pl:                 # uses __iter__
    print(f"  🎵 {song}")
````
