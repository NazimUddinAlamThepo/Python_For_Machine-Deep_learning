# Concatenation

Row wise concatenation -> column same thakte hobe


Column wise concatenation -> row same thakte hobe

```python
# Axis 0 → vertical direction ↓

# Axis 1 → horizontal direction →
```

```python
# Row wise

ans1 = np.concatenate((a,b) , axis =0) # axis 0 -> row wise

print(ans1)

print("=="*40)

# Column Wise

ans2 = np.concatenate((a,b) , axis = 1) # axis 1 -> column wise


print(ans2)
```

# Sorting

```python
# Two kind (Inplace , copy)


# inplace root change

arr = np.array([3,0,71,1,2])

arr.sort()

print(arr)
```

```python
# Copy sorting

arr = np.array([3,0,71,1,2])

sorted = np.sort(arr)

print(arr)
print(sorted)
```

```python
# Axis 0 → vertical direction ↓

# Axis 1 → horizontal direction →
```

```python
# 2D array sorting

import numpy as np

a = np.array([[23, 4, 345],
              [34, 1, 0]])

 # basic = np.sort(a) error asbe axis dite hobe

 # Sorting along axis=0 (column-wise)  vertical direction

col_sort = np.sort(a, axis=0)

row_sort = np.sort(a , axis=1)

print(col_sort)
print("=="*40)
print(row_sort)
```

# Searching

```python
# np.where(condition , x, y ) if true then x otherwise y

a = np.array([1,2,3,4,5,6,77,34,23,5,6,7,7,75,4,33,3,2,22,2,22])

index = np.where(a==5)
#index = np.where(a>5)

print(index)

index = np.where(a==5 , a , 0) # a thakle a i rakho otherwise change to 0

print(index)
```

```python
# 2D array searching

import numpy as np

arr = np.random.randint(1,10,(4,4))
print(arr)


result = np.where(arr>8)

print(result) # first array ta dibe all the row value
              # second array ta dibe all the column value
              # thats why duita array tuple akare dibe


print(arr[1][2])


result = np.where(arr>8 , arr, 0)
print(result)
```

```python
# Counting

# age hocche matrix take flattern kore then max index ta dey

max_index = np.argmax(arr)
print(max_index)
```

```python
# this will give row and column both at a time

pos = np.unravel_index(np.argmax(arr), arr.shape)
print(pos)
```
