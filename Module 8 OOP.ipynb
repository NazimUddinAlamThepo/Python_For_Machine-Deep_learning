```python
class Phone:
    manufactured = "China"

    # constructor
    def __init__(self, model, battery, camera, battery_percentage=100):
        self.model = model
        self.battery = battery
        self.camera = camera
        self.battery_percentage = battery_percentage

    def charge(self, charge_minutes):
        # Each minute = 1% charge
        self.battery_percentage += charge_minutes

        # Battery cannot exceed 100%
        if self.battery_percentage > 100:
            self.battery_percentage = 100

    def photo_capture(self, number_of_photos):
        # Each photo = -1% battery
        if self.battery_percentage > 0:
            self.battery_percentage -= number_of_photos

            # Battery cannot go negative
            if self.battery_percentage < 0:
                self.battery_percentage = 0
        else:
            print("Low battery")
```

```python
iphone = Phone(17 , 345 , "40MP" )
print(iphone.battery_percentage)

iphone.photo_capture(1000)
print(iphone.battery_percentage)

iphone.photo_capture(1000)
print(iphone.battery_percentage)
```

```python
# If we inherit multiple class that time we can't use super

class A:
    def __init__(self, name):
        self.name = name
        print(f"A constructor → name = {self.name}")


class B:
    def __init__(self, age):
        self.age = age
        print(f"B constructor → age = {self.age}")


class C(A, B):   # Multiple inheritance
    def __init__(self, name, age, city):

        A.__init__(self, name)   # directly call parent constructor
        # As usual we dont pass the "self" but in this case we actually pass the "self"

        B.__init__(self, age)    # directly call parent constructor

        self.city = city
        print(f"C constructor → city = {self.city}")


# Create object
obj = C("Rahim", 22, "Dhaka")
```

# **Polymorphism**

```python
# Overriding

class Dog:
    def speak(self):
        return "Woof!"

class Cat:
    def speak(self):
        return "Meow!"

class Cow:
    def speak(self):
        return "Moo!"

# Polymorphism
animals = [Dog(), Cat(), Cow()]

for animal in animals:
    print(animal.speak())

```

```python
class Shape:
    def area(self):
        return "Area not defined"

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side ** 2


# Polymorphism in action
shapes = [
    Circle(5),
    Rectangle(4, 6),
    Square(3)
]

for shape in shapes:
    print(shape.area())
```

```python
# ABC = Abstract Base Class
# ABC + abstractmethod = a way to force child classes to implement required functions.

from abc import ABC, abstractmethod

class Vehicle(ABC):

    @abstractmethod
    # abstractmethod = A method that must be overridden in child classes.
    def start(self):
        pass

    @abstractmethod
    def stop(self):
        pass

    def info(self):
        return "This is a vehicle"

# --------------------

class Car(Vehicle):
    def start(self):
        return "Car engine started"

    def stop(self):
        return "Car engine stopped"

    def info(self):  # overriding (polymorphism)
        return "This is a car"


# --------------------

class Bike(Vehicle):
    def start(self):
        return "Bike started with kick"

    def stop(self):
        return "Bike stopped"

    def info(self):  # overriding (polymorphism)
        return "This is a bike"


# --------------------

class Bus(Vehicle):
    def start(self):
        return "Bus engine roars to life"

    def stop(self):
        return "Bus engine shuts down"

    def info(self):
        return "This is a bus"

# --------------------

# Polymorphism in action
vehicles = [Car(), Bike(), Bus()]

for v in vehicles:
    print(v.info())
    print(v.start())
    print(v.stop())
    print()
```

# **Encapsulation**

```python
# Private  ( __ double underscore )
class A :
  def __init__(self , a , b):
    self.a = a
    self.__b = b     #Private

  def method1(self):
    print("From Class A Method 1")

  def getter(self):
    return self.__b

  def setter(self,b):
    self.__b = b


objectA = A(1,2)
print(objectA.getter())
objectA.setter(34)
print(objectA.getter())

```

```python
# protected ( _ single underscore)

from abc import ABC, abstractmethod

class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass

    # Protected abstract method
    @abstractmethod
    def _move(self):
        pass


class Dog(Animal):

    def sound(self):
        print("Dog barks")

    # implementing protected abstract method
    def _move(self):
        print("Dog runs")


# Testing
d = Dog()
d.sound()
d._move()   # allowed but discouraged (protected)
```
