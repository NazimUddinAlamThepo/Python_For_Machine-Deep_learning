🌟 Why NumPy Is Faster Than Python Lists

NumPy is built on a powerful core structure called the ndarray (N-Dimensional Array).
Understanding this explains why NumPy is much faster than Python lists.

🔷 1. Python List vs NumPy ndarray
❌ Python List (Slow)

List elements can be different data types.

Each element has its own separate memory pointer.

Memory is scattered, not continuous.

CPU must jump around memory, making access slow.

Example problem:

[5, "hello", 10.5, True]


Each item:

Different size

Different data type

Stored in different memory locations



🔷 2. NumPy ndarray (Fast)
✔ Same Data Type

All elements must be of the same type (all int, all float, etc.).

✔ Continuous Memory Block

Data is stored in one continuous block in RAM.

This results in:

Faster memory access

Faster computation

Better performance in loops and math operations

✔ Quick Comparison
Feature	Python List	NumPy ndarray
Data Types	Mixed	Same
Memory Layout	Scattered	Continuous
Speed	Slow	Fast



🔷 3. Vectorization = Huge Speed Boost

NumPy does operations on entire arrays at once, not one-by-one.

Example:

a = np.array([1, 2, 3, 4])

a * 2


NumPy applies multiplication to the whole array using one optimized instruction.



🔷 4. CPU Optimization: SIMD

Modern CPUs support SIMD (Single Instruction Multiple Data).

🧠 SIMD means:

CPU loads multiple values at the same time

Performs the same operation on them in parallel

Greatly improves performance

✔ NumPy ndarray

Works perfectly with SIMD because memory is continuous.

❌ Python list

CPU must load each element one at a time → slow.





✅ Final Summary

NumPy is fast because of the ndarray structure.

ndarray uses same data type + continuous memory.

Enables vectorized computation.

CPU uses SIMD parallel processing.

Python lists cannot do this → slower performance.

```python
# Slicing

# 1D array  -> arr[start : end : step]

# 2D array  -> arr[ row_start : row_end : step_for_row , column_start : column_end : step_for_column]


# getting a row

row_0 = arr[0:1 , ]

row_1 = arr[1: , ]


# getting column

column_0 = arr[0: , 0:1]

column_1 = arr[0: , 1:2]

print(column_1)

```

# Advance Indexing and Iteration

```python
# normal slicing is a view not a copy
# so if i change the view is also change the main array

arr = np.random.randint(1,5,(6,))

arr_slice_copy = arr[1:5]

arr_slice_copy[2]=100

print(arr_slice_copy)
print(arr)
```

```python
# solve using .copy()

arr = np.random.randint(1,5,(6,))

arr_slice_copy = arr[1:5].copy()

arr_slice_copy[2]=100

print(arr_slice_copy)
print(arr)
```
