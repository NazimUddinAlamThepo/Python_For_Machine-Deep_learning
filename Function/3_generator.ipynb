### Generator (iterator make korar jonno use kora hoy)

```python
def numbers():
    yield 1
    yield 2
    yield 3
    yield 4
    yield 5

x = numbers()
print(x)

print(next(x))
print(next(x))
print(next(x))
```

```python
# Actual iterator making 

def numbers():
    for i in range(5):
        yield i

x = numbers()
print(x)

print(next(x))
print(next(x))
print(next(x))
```

```python
# 10 GB --> 4 GB RAM
#passing value by chunk by chunk

lst = [ x for x in range(3444)]
#print(lst)

def loader(chunk_size , l):

  for i in range(0 , len(lst) ,chunk_size) :
    # i  = 0 , 5, 10
    yield l[i : i+chunk_size]

x = loader(5,lst)

print(x)
print(next(x))
print(next(x))
print(next(x))
```
