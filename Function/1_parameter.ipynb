### positional argument

```python
def square_addition(a,b,c):
    print(f"{a} {b} {c}")
    a=a**2 
    b= b**2 
    return a+b+c


#positional argument
ans = square_addition(10,20,30) + 10 # return na thakle ei 10 add kora jabe na
print(ans) 
```

### keyword argument

```python
#keyword argument

ans = square_addition(b=10,c=20,a=30) + 10 
print(ans) 
```

### Variable size arguments

```python
# use case calculator 

def sum(*args):

    result = 0
    
    for i in args:
        result += i;
    
    return result

```

```python
sum(12,12,14,15,16,17)
```

### Key value pair Arguments

```python
def student(   **      keywordarguments): 

    print("Full :", keywordarguments)

    
    for key,val in keywordarguments.items():
        print(f"{key} : {val}")



student(name = "adil" , cls =10 , roll=30 , marks = 30.5) 
```

### Return tuple

```python
def give_prediction(a,b):
    return a,b 

print(type(give_prediction(10,30)))
print(give_prediction(10,30))


## unpacking 
x , y = give_prediction(10,30)

print(x,y)
```

```python
def give_prediction2():

    a= {10,20,30}
    b = [100,200,300]
    return a,b 


def give_prediction1():
    return { "name" : "adil" , "addr" : "Ctg"},  { "name" : "abrar" , "addr" : "dhaka"}

p = give_prediction1()

x,y = give_prediction1()

z = give_prediction2()



#iteration in dictionary
for item in x.items():
    print(item) 

# iteration in tuple
for item in p:
    print(item) 

for item in z:
    print(item) 
```
