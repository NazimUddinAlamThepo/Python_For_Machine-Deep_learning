### positional argument

```python
def square_addition(a,b,c):
    print(f"{a} {b} {c}")
    a=a**2 
    b= b**2 
    return a+b+c


#positional argument
ans = square_addition(10,20,30) + 10 # return na thakle ei 10 add kora jabe na
print(ans) 
```

### keyword argument

```python
#keyword argument

ans = square_addition(b=10,c=20,a=30) + 10 
print(ans) 
```

### Variable size arguments

```python
# use case calculator 

def sum(*args):

    result = 0
    
    for i in args:
        result += i;
    
    return result

```

```python
sum(12,12,14,15,16,17)
```
