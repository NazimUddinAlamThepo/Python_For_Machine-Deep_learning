# Map Function

```python
# prottekti vlaue er upor ekti nirdisto function apply korte pare

# map (je function ta sobar jonno use korte chacchi seta , jar upor apply korte chacchi)


lst = [1,23,4,5,6,7]

def square(x):
  return x**2

all_square = map(square,lst)

print(all_square)

all_square = list(map(square,lst))
print(all_square)
```

```python
text = "AMhias dasd sdsdfdsfsd fsdfsd fsdf sdfsdf sdfsd fdfsdf sdfsdf sdf dsfsdfsdf sdfsdfsdf sdf"

h = str.upper(text)
print(h)

# map applies the function to each character of the string
mapped = map(str.upper, text)   # function ta call na just method ta object pathate hoy str.upper()

# convert to list
char_list = list(mapped)
print(char_list)

# join list back to string
result = ''.join(char_list)
print(result)
```

```python
# map with lambda

lst = [1,2,3,4,5,6,7]

lst = map( lambda x : x**2 ,lst)

lst = list(lst)
print(lst)
```

```python
# if else is one statement by using this we can make a lamda function

lst = [1,2,3,4,5,6,7]

y  =  lambda x : x**2 if x%2 == 0 else x

print(y(3))
```

```python
lst = [1,2,3,4,5,6,7]

lst = map( lambda x : x**2 if x%2 == 0 else x ,lst)

lst = list(lst)
print(lst)
```

```python
# map with lambda for finding even

lst = [1,2,3,4,5,6,7]

lst = map( lambda x : x if x%2 ==0 else None ,lst)

lst = list(lst)
print(lst)
```

```python
lst = [1, 2, 3, 4, 5, 6, 7]

lst = map(lambda x: x % 2 == 0, lst)

lst = list(lst)
print(lst)
```

# Filter function

```python
# filter ( true/false function , jar upor apply korbo )

numbers = [x for x in range(34)]

# even = filter("True false Fuction" , " jar upor apply korbo")

even = filter(lambda x : x%2==0 , numbers)
even = list(even)
print(even)


# greater than 20

up = list(filter(lambda x : x>20 , numbers))
print(up)
```

```python
mixed_list = [
    None,                  # starts with None
    42,                    # integer
    3.14,                  # float
    None,                  # another None
    "Hello, World!",       # string
    True,                  # boolean
    [1, 2, None, 3],       # list containing None
    (4, None, 6),          # tuple containing None
    {"name": "Mahi", "age": None},  # dictionary with None value
    None,                  # another None
    complex(2, 3)          # complex number
]

cleaned_data = list(filter(None , mixed_list))   # ulta kaj kore none dichi none re bad diya baki gula diche

print(cleaned_data)


```

```python
long_text = """
The sun rises in the east and sets in the west.
Nature has its own rhythm — from the rustle of leaves
to the calm waves touching the shore.

Life is full of patterns, repetitions, and reflections.
We learn, forget, and learn again.
None of it is wasted — every moment shapes us.

Repeat after me: consistency is key.
Repeat after me: patience is power.
Repeat after me: effort always counts.
"""


vow = list(filter(lambda x : x in "AEIOUaeiou" , long_text))
vow = ''.join(vow)
print(vow)
```

```python
string_list = [
    "Hi",                                 # length 2
    "Hello",                              # length 5
    "Python",                             # length 6
    "Data Science",                        # length 12
    "AI",                                  # length 2
    "Machine Learning is amazing",         # length 27
    "OpenAI",                              # length 6
    "I love coding in Python",             # length 23
    "GPT",                                 # length 3
    "This is a very long string for testing purposes"  # length 49
]

long_length_string = list(filter(lambda x : len(x) >20 , string_list))
print(long_length_string)
```

# Reduce Funtion From functool library

```python
# কোন একটা iterable  থেকে আমাদের define logic  এর মাধ্যমে একটা result দিবে ( রেজাল্ট দিবে মাত্র একটা )

# Structure as like as the map function, filter function

from functools import reduce
```

```python
list = [1,2,3,4,5,6,7,8]

sum = reduce(lambda x,y : x+y  , list)  # No type conversion needed

print(sum)


max = reduce (lambda x,y : x if x>y else y , list)
print(max)
```

```python
string_list = [
    "Hi",
    "Hello",
    "Python",
    "Data",
    "AI",
]

# String conversion using reduce function

string = reduce (lambda x ,y : x+ " " + y , string_list)
print(string)
```
